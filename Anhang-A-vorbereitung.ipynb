### Achtung: „3D“ bei Tensoren ≠ „3D“ in der Mathematik

Bei Tensoren zählt **D** die **Achsen** (`tensor.ndim`), also wie viele Indizes man braucht, um einen einzelnen Wert anzusprechen. Wie viele Werte auf einer Achse liegen, spielt dafür keine Rolle.

| Begriff | Bedeutung | `shape` | `ndim` |
|---|---|---|---|
| 3D-Vektor | 3 Komponenten, z. B. ein Punkt im Raum (x, y, z) | `[3]` | **1** |
| 3D-Matrix (3×3) | 3 Zeilen × 3 Spalten, z. B. eine Rotation im Raum | `[3, 3]` | **2** |
| 3D-Tensor | Werte entlang 3 Achsen, z. B. `torch.zeros(2, 3, 4)` | `[2, 3, 4]` | **3** |

In der Mathematik meint „3D“ also den **Raum mit 3 Richtungen**, bei Tensoren die **Anzahl der Achsen**.

Faustregel: `ndim` = Länge von `shape` = Anzahl der öffnenden eckigen Klammern am Anfang der Ausgabe. Die Zahlen *in* `shape` geben an, wie lang jede Achse ist.

```python
vektor_3d = torch.tensor([1.0, 2.0, 3.0])  # 3D-Vektor aus der Mathematik
matrix_3x3 = torch.eye(3)                  # 3×3-Matrix („3D-Matrix“)
tensor_3d = torch.zeros(2, 3, 4)           # 3D-Tensor

for name, t in [("3D-Vektor", vektor_3d), ("3×3-Matrix", matrix_3x3), ("3D-Tensor", tensor_3d)]:
    print(f"{name:<11} shape = {list(t.shape)},  ndim = {t.ndim}")
    print(t)
```

## Automatische Ableitung mit Autograd

Gradienten des Verlusts nach den Parametern `w1` und `b` berechnen.

```python
import torch
import torch.nn.functional as F
from torch.autograd import grad

y = torch.tensor([1.0])
x1 = torch.tensor([1.1])
w1 = torch.tensor([2.2], requires_grad=True)
b = torch.tensor([-0.5], requires_grad=True)
z = x1 * w1 + b
a = torch.sigmoid(z)
loss = F.binary_cross_entropy(a, y)

print('z: ', z, ' a: ', a, ' loss: ', loss)

grad_L_w1 = grad(loss, w1, retain_graph=True)
grad_L_b = grad(loss, b, retain_graph=True)

print(grad_L_w1)
print(grad_L_b)

# Even easier directly
loss.backward()
print(w1.grad)
print(b.grad)

```

## Simple Neural Network

```python
class NeuralNetwork(torch.nn.Module):
    def __init__(self, num_inputs, num_outputs):
        super().__init__()

        self.layers = torch.nn.Sequential(
            # 1st hidden layer
            torch.nn.Linear(num_inputs, 30),
            torch.nn.ReLU(),

            # 2nd hidden layer
            torch.nn.Linear(30, 20),
            torch.nn.ReLU(),

            # output layer
            torch.nn.Linear(20, num_outputs),
        )

    def forward(self, x):
        logits = self.layers(x)
        return logits


```

```python
model = NeuralNetwork(50, 3)

print(model)
```

```python
num_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print("Total number of trainable model parameters:", num_params)
```

```python
print(model.layers[0].weight)
```

```python
print(model.layers[0].weight.shape)
```

## Trainingsdaten vorbereiten

```python
X_train = torch.tensor([
    [-1.2, 3.1],
    [-0.9, 2.9],
    [-0.5, 2.6],
    [2.3, -1.1],
    [2.7, -1.5]
])
y_train = torch.tensor([0, 0, 0, 1, 1])

X_test = torch.tensor([
    [-0.8, 2.8],
    [2.6, -1.6],
])
y_test = torch.tensor([0, 1])
```

Man muss eine Dataset Klasse manuell implementieren, um die Daten in einem DataLoader zu verwenden. Die Klasse muss die Methoden `__len__` und `__getitem__` implementieren.

```python
from torch.utils.data import Dataset

class ToyDataset(Dataset):
    def __init__(self, X, y):
        self.features = X
        self.labels = y

    def __getitem__(self, index):
        one_x = self.features[index]
        one_y = self.labels[index]
        return one_x, one_y

    def __len__(self):
        return self.labels.shape[0]

train_ds = ToyDataset(X_train, y_train)
test_ds = ToyDataset(X_test, y_test)

print(len(train_ds))
```

DataLoader initialisieren mit unseren Datasets.

```python
from torch.utils.data import DataLoader

torch.manual_seed(123)

train_loader = DataLoader(
    dataset=train_ds,
    batch_size=2,
    shuffle=True,
    num_workers=0,
        drop_last=True # This drops the last possibly uneven dataset to not distort the training result

)

test_loader = DataLoader(
    dataset=test_ds,
    batch_size=2,
    shuffle=False,
    num_workers=0,
)
```

Zeige die vorbereiteten Trainingsdaten:

```python
for idx, (x, y) in enumerate(train_loader):
    print(f"Batch {idx}:", x,y)
```

## Training eines NN durchführen

```python
import torch.nn.functional as F

torch.manual_seed(123)
model = NeuralNetwork(num_inputs=2, num_outputs=2)
optimizer = torch.optim.SGD(
    model.parameters(), lr=0.5
)

num_epochs = 3

for epoch in range(num_epochs):
    model.train()

    for batch_idx, (features, labels) in enumerate(train_loader):
        logits = model(features)

        loss = F.cross_entropy(logits, labels)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        ### LOGGING
        print(f"Epoch {epoch+1:03d}/{num_epochs:03d}"
              f" | Batch  {batch_idx+1:03d}/{len(train_loader):03d}"
              f" | Train Loss: {loss:.2f}")

model.eval()
```

Define a function to do the above accuracy measurement

```python
def compute_accuracy(model, dataloader):
    model = model.eval()
    correct = 0.0
    total_examples = 0

    for idx, (features, labels) in enumerate(dataloader):
        with torch.no_grad():
            logits = model(features)

        predictions = torch.argmax(logits, dim=1)
        compare = labels == predictions
        correct += torch.sum(compare)
        total_examples += len(compare)

    return (correct / total_examples).item()

```

```python
print(compute_accuracy(model, train_loader))

print(compute_accuracy(model, test_loader))
```

Übung: Herausfinden, ab welcher Batchgröße die GPU schneller ist als die CPU.

Ab welcher Matrixgröße ist die GPU (MPS) schneller als die CPU? Gemessen wird eine Matrixmultiplikation `a @ b` mit zwei quadratischen Zufallsmatrizen. Verwendet `device` aus dem Trainings-Code oben.

GPU-Aufrufe laufen asynchron, deshalb wird vor dem Stoppen der Zeit mit `synchronize` auf das Ende gewartet. Der erste Aufruf (Warm-up) wird nicht gemessen.

```python
import time

def synchronize(device):
    # Wartet, bis alle Berechnungen auf der GPU fertig sind
    if device.type == "cuda":
        torch.cuda.synchronize()
    elif device.type == "mps":
        torch.mps.synchronize()

def time_matmul(size, device, repeats=10):
    torch.manual_seed(123)
    a = torch.rand((size, size)).to(device)
    b = torch.rand((size, size)).to(device)

    a @ b  # Warm-up
    synchronize(device)

    start = time.perf_counter()
    for _ in range(repeats):
        a @ b
    synchronize(device)
    return (time.perf_counter() - start) / repeats

sizes = [8, 16, 32, 64, 128, 256, 512, 768, 1024, 1536, 2048, 3072, 4096]
crossover = None

print(f"Vergleich cpu vs. {device}\n")
print(f"{'Größe':>6} | {'CPU (ms)':>10} | {'GPU (ms)':>10} | Schneller")
for size in sizes:
    t_cpu = time_matmul(size, torch.device("cpu"))
    t_gpu = time_matmul(size, device)
    faster = "GPU" if t_gpu < t_cpu else "CPU"
    if faster == "GPU" and crossover is None:
        crossover = size
    print(f"{size:>6} | {t_cpu*1000:>10.3f} | {t_gpu*1000:>10.3f} | {faster}")

if crossover:
    print(f"\nDie GPU ist ab Matrixgröße {crossover}x{crossover} schneller.")
else:
    print("\nDie GPU war bei keiner getesteten Größe schneller.")
```
